## Why Check Existence First?

Checking whether a file exists helps students debug path mistakes before trying to load the file.

That means fewer crashes and fewer emotional speeches directed at the laptop.

````python
possible_files = [
    Path("data/countries.geojson"),
    Path("data/missing.geojson"),
    Path("output"),
]

for p in possible_files:
    print(f"{p!s:<28} exists={p.exists()}  is_file={p.is_file()}  is_dir={p.is_dir()}")
````

## Exercise A

In your own words, describe the difference between these three methods. For each one, state what it returns when given: a path to a real file, a path to a real directory, and a path to something that doesn't exist at all.

- `.exists()`
- `.is_file()`
- `.is_dir()`


- **.exists()**: This method determines if a specified path is present. It will return True for actual files and directories, and False if the path doesn't exist.
- **.is_file()**: This checks specifically if the path refers to a file, returning True only for files and False for directories or paths that are missing.
- **.is_dir()**: This function verifies if the path is a directory, returning True solely for directories and False if it's a file or if the path is not found.

## Optional Advanced — Set Up a Local Workspace

Use `pathlib` (no shell commands) to:

1. Create a `data/` folder inside `00-Paths/` using `Path.mkdir(exist_ok=True)`
2. Create an `output/` folder the same way
3. Copy one json file from the module's data folder into your new local `data/` using `shutil.copy2`
4. Re-run the existence check from above and confirm all three paths are found

Hint: `Path.mkdir(parents=True, exist_ok=True)` will not error if the folder already exists.

````python
from pathlib import Path

# Create a local data/ and output/ folder inside 00-Paths/ if they don't exist
# Then copy one json file from the module data folder into data/
# Finally, re-run the existence check from above to confirm everything is found

# Your code here
from pathlib import Path
import shutil

# 1. Create folders inside current working directory
data_dir = Path("data")
output_dir = Path("output")

data_dir.mkdir(parents=True, exist_ok=True)
output_dir.mkdir(parents=True, exist_ok=True)

# 2. Copy uploaded file into data/
source = Path("countries.geojson")
destination = data_dir / "countries.geojson"

shutil.copy2(source, destination)

# 3. Confirm everything exists
check_paths = [
    data_dir,
    output_dir,
    destination
]

for p in check_paths:
    print(f"{p} exists={p.exists()}")
````
