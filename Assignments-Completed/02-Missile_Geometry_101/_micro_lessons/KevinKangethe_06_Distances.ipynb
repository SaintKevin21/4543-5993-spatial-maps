```python
from pathlib import Path


## Use last items learned from last lesson to locate and check if the file exists
## Remember parent, and folder: data

# ----------------------------------------
# Distance: Haversine (km)
# ----------------------------------------
def haversine_km(lat1: float, lon1: float, lat2: float, lon2: float) -> float:
    """
    Compute great-circle distance between two lat/lon points.
    Returns distance in kilometers.
    Inputs: degrees
    Output: kilometers
    """
    phi1 = radians(lat1)
    phi2 = radians(lat2)

    dphi = radians(lat2 - lat1)
    dlmb = radians(lon2 - lon1)

    a = sin(dphi / 2) ** 2 + cos(phi1) * cos(phi2) * sin(dlmb / 2) ** 2
    c = 2 * atan2(sqrt(a), sqrt(1 - a))

    return EARTH_RADIUS_KM * c



debug = False

cwd = Path.cwd()
if debug:
    print("Current Working Directory:")
    print(cwd)

target = cwd / "countries.geojson"
if debug:
    print("\nAttempting to access:")
    print(target)

exists = target.exists()
if debug:
    print("\nDoes file exist?")
    print(exists)

assert exists, (
    f"\n❌ ERROR: File not found:\n{target}\n"
    "Check spelling and folder structure."
)

```

```python
from google.colab import files
import folium
import json
from pathlib import Path
from math import radians, sin, cos, sqrt, atan2
import random

# ----------------------------------------
# Upload file
# ----------------------------------------
uploaded = files.upload()

# ----------------------------------------
# Load data
# ----------------------------------------
with open("world_cities_fixed.json", "r", encoding="utf-8") as f:
    data = json.load(f)

print("Cities loaded:", len(data))

# ----------------------------------------
# MSU location (Wichita Falls area)
# ----------------------------------------
msu = (33.8634, -98.5210)

# ----------------------------------------
# Distance function
# ----------------------------------------
EARTH_RADIUS_KM = 6371.0

def distance(lat1, lon1, lat2, lon2):
    phi1, phi2 = radians(lat1), radians(lat2)
    dphi = radians(lat2 - lat1)
    dlmb = radians(lon2 - lon1)

    a = sin(dphi/2)**2 + cos(phi1)*cos(phi2)*sin(dlmb/2)**2
    c = 2 * atan2(sqrt(a), sqrt(1-a))
    return EARTH_RADIUS_KM * c

# ----------------------------------------
# Create map
# ----------------------------------------
m = folium.Map(location=msu, zoom_start=3)

# MSU marker
folium.Marker(msu, popup="MSU (Wichita Falls)", icon=folium.Icon(color="red")).add_to(m)

# ----------------------------------------
# Sample cities (IMPORTANT for performance)
# ----------------------------------------
sample_cities = random.sample(data, 200)

# ----------------------------------------
# Draw lines
# ----------------------------------------
for city in sample_cities:

    try:
        name = city["city-name"]
        lat = float(city["lat"])
        lon = float(city["lon"])

        dist = distance(msu[0], msu[1], lat, lon)

        # filter (optional but improves clarity)
        if dist > 8000:
            continue

        folium.PolyLine(
            locations=[msu, (lat, lon)],
            popup=f"{name} - {dist:.1f} km",
            weight=1
        ).add_to(m)

    except:
        continue

# ----------------------------------------
# Save map
# ----------------------------------------
m.save("msu_city_lines.html")

print("Map saved as msu_city_lines.html")
```
